Use multiple built-in functions together to process a list of numbers

```python
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

positive_numbers = list(filter(lambda x: x > 0, numbers))

print("Positive numbers:", positive_numbers)
print("Total:", sum(positive_numbers))
print("Maximum:", max(positive_numbers))
print("Minimum:", min(positive_numbers))
print("Count:", len(positive_numbers))
```
